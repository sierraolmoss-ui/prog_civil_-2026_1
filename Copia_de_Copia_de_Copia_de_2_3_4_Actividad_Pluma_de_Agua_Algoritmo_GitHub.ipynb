## 2. Situación problema

Un hogar recibe una cierta cantidad de **plumas de agua**.  
Se sabe que:

- **1 pluma = 0.025 litros por segundo**

Con esta información, el estudiante debe construir un programa que permita calcular:

1. el caudal total en litros por segundo;
2. el consumo diario en litros;
3. el consumo mensual en litros (suponiendo 30 días);
4. una clasificación del consumo:
   - **Bajo**: menos de 1000 litros por día
   - **Medio**: entre 1000 y 3000 litros por día
   - **Alto**: más de 3000 litros por día

Complete el siguiente bloque de código a partir del algoritmo anterior.

````python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# ENTRADA :Solicitar al usuario el número de plumas
plumas = float(input("Ingrese el número de plumas: "))

# PROCESO : Calcular el caudal total , consumo diario y consumo mensual

##Definir constantes del problema

Caudal_pluma= 0.025 # litros por segundo

segundos_dia= 24 * 60 * 60 # segundos

dias_mes= 30 # dias

##Calcular el caudal total de agua que fluye a través de todas las plumas asignadas.

caudal_total= plumas * Caudal_pluma # litros/segundo

##Calcular el consumo diario

consumo_diario= caudal_total * segundos_dia # litros/dia

##Calcular el consumo mensual

consumo_mensual= consumo_diario * dias_mes # litros/mes

## Clasificar el consumo
nivel_consumo=""

if consumo_diario < 1000:
  nivel_consumo = "Bajo"
elif 1000 <= consumo_diario <= 3000:
  nivel_consumo = "Medio"
else:
  nivel_consumo = "Alto"

# SALIDA: imprimir el resultado

## Mostrar resultados
print(f" Resultados del Consumo de Agua ")
print(f"Número de plumas ingresadas: {plumas}")
print(f"Caudal total: {caudal_total:.3f} litros por segundo")
print(f"Consumo diario: {consumo_diario:.2f} litros")
print(f"Consumo mensual: {consumo_mensual:.2f} litros")
print(f"Clasificación del consumo: {nivel_consumo}")
````

### 6. Versión Comentada del Código (Sin Cambiar su Lógica)

```python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# ENTRADA: Solicitar al usuario el número de plumas
# Se pide al usuario que ingrese la cantidad de 'plumas de agua' que tiene un hogar.
# La entrada se convierte inmediatamente a tipo float para permitir valores decimales.
plumas = float(input("Ingrese el número de plumas: "))

# PROCESO: Calcular el caudal total, consumo diario y consumo mensual, y clasificar el consumo.

## Definir constantes del problema
# Caudal de una única pluma en litros por segundo (L/s).
Caudal_pluma = 0.025
# Cantidad total de segundos en un día (24 horas * 60 minutos/hora * 60 segundos/minuto).
segundos_dia = 24 * 60 * 60
# Número de días considerado para el cálculo del consumo mensual.
dias_mes = 30

## Calcular el caudal total de agua que fluye a través de todas las plumas asignadas.
# Multiplica el número de plumas ingresado por el caudal de una pluma individual.
caudal_total = plumas * Caudal_pluma # litros/segundo

## Calcular el consumo diario
# Convierte el caudal total (L/s) a consumo diario (L/día) multiplicando por los segundos en un día.
consumo_diario = caudal_total * segundos_dia # litros/dia

## Calcular el consumo mensual
# Calcula el consumo total en litros para un mes, basado en el consumo diario y los días del mes.
consumo_mensual = consumo_diario * dias_mes # litros/mes

## Clasificar el consumo
# Inicializa la variable que contendrá la clasificación del consumo.
nivel_consumo = ""

# Utiliza una estructura condicional if-elif-else para clasificar el consumo diario.
# Si el consumo diario es menor de 1000 litros.
if consumo_diario < 1000:
  nivel_consumo = "Bajo"
# Si el consumo diario está entre 1000 y 3000 litros (inclusive).
elif 1000 <= consumo_diario <= 3000:
  nivel_consumo = "Medio"
# Si el consumo diario es mayor de 3000 litros.
else:
  nivel_consumo = "Alto"

# SALIDA: Imprimir los resultados calculados al usuario.

## Mostrar resultados
# Imprime un encabezado para los resultados.
print(f" Resultados del Consumo de Agua ")
# Muestra el número de plumas ingresado por el usuario.
print(f"Número de plumas ingresadas: {plumas}")
# Muestra el caudal total, formateado a tres decimales.
print(f"Caudal total: {caudal_total:.3f} litros por segundo")
# Muestra el consumo diario, formateado a dos decimales.
print(f"Consumo diario: {consumo_diario:.2f} litros")
# Muestra el consumo mensual, formateado a dos decimales.
print(f"Consumo mensual: {consumo_mensual:.2f} litros")
# Muestra la clasificación del consumo determinada.
print(f"Clasificación del consumo: {nivel_consumo}")
```
