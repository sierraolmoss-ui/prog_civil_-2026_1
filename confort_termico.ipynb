## Teoría del Confort Térmico en Ganado Vacuno

El **confort térmico** en el ganado vacuno se refiere al rango de temperaturas ambientales en el que los animales pueden mantener su temperatura corporal central sin tener que activar mecanismos reguladores de calor excesivos (como el jadeo o el aumento de la sudoración) o de frío (como el temblor o el aumento del consumo de alimento para producir calor). Cuando los animales se encuentran fuera de este rango de confort, experimentan **estrés térmico**.

### Rango de Termoneutralidad (RTN)
El RTN es el rango de temperatura ambiente en el que los animales no necesitan gastar energía adicional para mantener su temperatura corporal. Dentro de este rango, la producción y pérdida de calor se equilibran de forma natural. Sin embargo, este rango varía según:

*   **Especie y Raza:** Algunas razas son más tolerantes al calor (p. ej., Cebú) que otras (p. ej., Bos taurus).
*   **Edad y Etapa Productiva:** Terneras, vacas lecheras de alta producción y animales en engorde son más susceptibles al estrés térmico.
*   **Nivel de Alimentación:** Una mayor ingesta de alimento genera más calor metabólico.
*   **Estado de Salud:** Animales enfermos son más vulnerables.

### Estrés por Calor
El estrés por calor ocurre cuando la temperatura ambiental y la humedad son tan altas que el animal no puede disipar el calor de su cuerpo de manera eficiente. Esto puede llevar a:

*   **Fisiología:** Aumento de la temperatura corporal, frecuencia respiratoria (jadeo), frecuencia cardíaca, sudoración.
*   **Producción:** Disminución en la producción de leche y carne, reducción de la tasa de crecimiento, menor eficiencia alimenticia, problemas reproductivos, debilitamiento del sistema inmune.
*   **Comportamiento:** Búsqueda de sombra, reducción de la ingesta de alimento (especialmente durante las horas más calurosas), aumento del consumo de agua.

### Estrés por Frío
Aunque menos común en climas tropicales, el estrés por frío se produce cuando las temperaturas son bajas y el animal debe aumentar su metabolismo para generar calor, lo que también impacta negativamente la producción y el bienestar.

### Índices de Confort Térmico
Para cuantificar y predecir el estrés térmico, se utilizan diversos índices que combinan factores ambientales. El **Índice de Temperatura y Humedad (ITH)** es uno de los más utilizados, ya que considera tanto la temperatura del aire como la humedad relativa, dos de los factores más influyentes en la sensación térmica de los animales.

Los umbrales de ITH permiten clasificar la situación en diferentes niveles (no estresante, alerta, peligro, crítica), lo que facilita la toma de decisiones para implementar estrategias de mitigación y adaptación al cambio climático, protegiendo así la productividad y el bienestar animal.

#**Métodos y cálculos**
+ Índice de temperatura y humedad (producción intensiva)

$$ ITH = 0.72(T_{bs} +T_{bh}) + 40.6 $$

Donde=

1. $ T_{bs} $ Temperatura bulbo seco ($°C $)

2. $ T_{bh} $ Temperatura de bulbo húmedo ($°C $)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda
utilizar los límites propuestos por National Weather Service:

1. Sí $ ITH <=70$ ambiente no estresante

2. Si $ ITH > 70 \space
and\ ITH <= 78$ situación de alerta por estrés térmico

3. $ Sí ITH > 78 \space and\ <= 83$ peligro por estrés térmico

4. Si $ITH > 83$ situación crítica


````python
# ENTRADA: pedir tbs (temperatura del bulbo seco) y tbh ( temperatura de bulbo humedo )
tbs = float(input("Ingrese la temperatura del bulbo seco, grados centigrados: "))
tbh = float(input("Ingrese la temperatura del bulbo húmedo, grados centigrados: "))

# PROCESO:calculo de ith(indice de temperatura y humedad)
ith = 0.72*(tbs + tbh) + 40.6

#SALIDA: imprimir resultado de la evaluacion de ith y el confort termico
if ith <= 70:
  print("el ambiente no es estresante")
elif ith > 70 and ith <= 78:
  print("la situacion de alerta por estrés térmico")
elif ith > 78 and ith <= 83:
  print("el peligro por estrés térmico")
else:
    print("situacion critica")
````
