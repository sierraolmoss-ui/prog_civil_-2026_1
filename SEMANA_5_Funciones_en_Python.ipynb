## SESIÓN 1: Introducción a las Funciones

**El Problema de la Repetición**
Supongamos que en un levantamiento topográfico necesitamos calcular el área de 50 terrenos rectangulares distintos. Escribir la operación `base * altura` 50 veces haría el programa largo y difícil de mantener.

**La Solución: Funciones**
Una función es un bloque de código encapsulado que realiza una tarea específica y que podemos "llamar" o ejecutar todas las veces que queramos.

### 1. Definición y Llamado de una Función Simple
Para crear una función en Python usamos la palabra clave `def`, seguida del nombre que queramos darle y paréntesis `()`. Todo lo que esté indentado (con un espacio hacia la derecha) pertenecerá a esa función.

### Celda 3: Código (Python)

````
# Definición de la función (El programa la aprende, pero aún no hace nada visible)
def mostrar_alerta_climatica():
    print("ALERTA: Se pronostican fuertes lluvias en la región de la sabana.")
    print("Revisar niveles de drenaje en los cultivos.")

# Llamado de la función (Aquí es donde realmente se ejecuta)
mostrar_alerta_climatica()
````

### 2. Funciones con Parámetros (Entradas)
Las funciones son mucho más útiles cuando pueden recibir datos para procesarlos. A estos datos los llamamos **parámetros**. Esto coincide perfectamente con nuestro modelo algorítmico de **Entrada $\rightarrow$ Proceso $\rightarrow$ Salida**.

### Celda 5: Código (Python)

````
# Función que recibe dos parámetros: base y altura
def calcular_area_terreno(base, altura):
    area = base * altura
    print("El área calculada es:", area, "metros cuadrados.")

# Reutilizamos la misma función con diferentes datos
print("Terreno 1:")
calcular_area_terreno(15.5, 20.0)

print("\nTerreno 2:")
calcular_area_terreno(8.0, 12.0)
````

### Mini-proyecto: Cálculo de Velocidad Media
En topografía o diseño de canales, calcular velocidades es rutinario ($v = \frac{d}{t}$). Ejecuta y analiza este código modular.

### Celda 10: Código (Python)

````
def velocidad_media(distancia, tiempo):
    v = distancia / tiempo
    return v

# Bloque principal del programa
d = float(input("Ingrese la distancia recorrida (m): "))
t = float(input("Ingrese el tiempo transcurrido (s): "))

vel_calculada = velocidad_media(d, t)
print("La velocidad media del flujo o vehículo es:", vel_calculada, "m/s")
````

# Problema: Número de Reynolds y Clasificación del Flujo

### Curso de Programación con Python — Tema: Funciones

---

## Contexto

En mecánica de fluidos, el **Número de Reynolds (Re)** es un parámetro adimensional que permite predecir el comportamiento de un fluido en movimiento dentro de una tubería. Su cálculo y la posterior clasificación del tipo de flujo son tareas fundamentales en ingeniería.

---

## Fundamento Teórico

### Número de Reynolds

$$Re = \frac{\rho \cdot v \cdot D}{\mu}$$

| Símbolo | Descripción | Unidades (SI) |
|--------|------------|--------------|
| $\rho$ | Densidad del fluido | kg/m³ |
| $v$ | Velocidad del fluido | m/s |
| $D$ | Diámetro de la tubería | m |
| $\mu$ | Viscosidad dinámica del fluido | Pa·s (kg/m·s) |
| $Re$ | Número de Reynolds | adimensional |

### Clasificación del Tipo de Flujo

$$\text{Tipo de flujo} = \begin{cases} \text{Laminar} & \text{si } Re < 2000 \\ \text{Transición} & \text{si } 2000 \leq Re \leq 4000 \\ \text{Turbulento} & \text{si } Re > 4000 \end{cases}$$

---

## Enunciado del Problema

Una empresa de ingeniería necesita una herramienta computacional para analizar el comportamiento de fluidos en tuberías. Desarrolle un programa en Python que, dados los parámetros de un fluido y una tubería, calcule el Número de Reynolds e indique el tipo de flujo resultante.

**Datos de prueba:**
- Densidad del fluido: $\rho = 1000 \text{ kg/m}^3$ (agua a 20°C)
- Velocidad del fluido: $v = 2.5 \text{ m/s}$
- Diámetro de la tubería: $D = 0.05 \text{ m}$
- Viscosidad dinámica: $\mu = 0.001 \text{ Pa·s}$

---

## Parte A — Solución con una sola función

Escriba **una única función** llamada `analizar_flujo` que reciba como parámetros $\rho$, $v$, $D$ y $\mu$, calcule internamente el Número de Reynolds, determine el tipo de flujo, y retorne **ambos resultados**.

Llame a la función con los datos de prueba e imprima los resultados con el siguiente formato:

```
Número de Reynolds: 125000.00
Tipo de flujo: Turbulento
```

> 💡 *Pista: una función puede retornar más de un valor usando* `return valor1, valor2`

---

## Parte B — Solución con dos funciones

Resuelva el mismo problema dividiendo la lógica en **dos funciones especializadas**:

**Función 1 — `calcular_reynolds`**
Recibe $\rho$, $v$, $D$ y $\mu$, calcula y **retorna** el valor numérico de $Re$.

**Función 2 — `clasificar_flujo`**
Recibe $\rho$, $v$, $D$ y $\mu$, **llama internamente a `calcular_reynolds`** para obtener $Re$, determina el tipo de flujo y retorna el valor de $Re$ junto con la clasificación.

Llame a `clasificar_flujo` con los datos de prueba e imprima los resultados con el mismo formato de la Parte A.

> **Nota:** El orden propuesto (Parte A → Parte B) es intencional: primero se consolida el concepto de función única para luego avanzar hacia la **modularidad y reutilización de código**, que es uno de los principios más importantes en programación.

````

"""
Parte A — Solución con una sola función
Escriba una única función llamada analizar_flujo que reciba como parámetros  ρ ,  v ,  D  y  μ ,
calcule internamente el Número de Reynolds, determine el tipo de flujo, y retorne ambos resultados.

Clasificación del Tipo de Flujo
Tipo de flujo=Laminar, Transición, Turbulentosi
Laminar = Re<2000
Transición = 2000≤Re≤4000
Turbulento = Re>4000
"""
def analizar_flujo(rho, v, D, mu):
    Re = (rho*v*D)/mu
    #Evalua el tipoo de flujo
    if Re < 2000:
        tipo_flujo = "Laminar"
    elif 2000 <= Re <= 4000:
        tipo_flujo = "Transición"

    else:
        tipo_flujo = "Turbulento"

    return Re, tipo_flujo

"""
Datos de prueba:

Densidad del fluido:  ρ=1000 kg/m3  (agua a 20°C)
Velocidad del fluido:  v=2.5 m/s
Diámetro de la tubería:  D=0.05 m
Viscosidad dinámica:  μ=0.001 Pa⋅s
"""

rho = float(input("Ingrese la densidad del fluido: "))
v = float(input("Ingrese la velocidad del fluido: "))
D = float(input("Ingrese el diámetro de la tubería: "))
mu = float(input("Ingrese la viscosidad dinámica: "))

Re, tipo_flujo = analizar_flujo(rho, v, D, mu)

print("Número de Reynolds:", Re)
print("Tipo de flujo:", tipo_flujo)
````
